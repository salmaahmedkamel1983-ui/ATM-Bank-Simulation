```python
# ==========================================
# مشروع محاكاة نظام بنكي بسيط (ATM Simulation)
# لغة البرمجة: Python
# الوصف: نظام محاكاة لعمليات صراف آلي (إيداع، سحب، استعلام) مع دعم اللغتين (عربي/إنجليزي)
# ==========================================

# نظام التحقق من البطاقة وكلمة المرور مع السماح بـ 3 محاولات فقط
for i in range(3):
    card_num=input("Enter your number's card :")
    password=input("Enter your password :")
    # التحقق من صحة بيانات البطاقة
    if card_num =="9876543" and password=="6789":
        # طلب إدخال الرصيد الابتدائي للحساب
        balance=int(input("ادخل رصيدك :"))
        # اختيار لغة الواجهة (عربي أو إنجليزي)
        choice=input("choose your language: \n 1.English \n 2. اللغة العربية \n")
        # ==========================================
        # واجهة اللغة العربية
        # ==========================================
        if choice=="2":
            print("اهلا بك")
            while True:

                operation=input("اختر العملية التي تريد القيام بها:\n 1-استعلام عن الرصيد \n 2- سحب نقدي \n 3-ايداع \n 4-خروج \n")
                # 1. الاستعلام عن الرصيد
                if operation == "1":
                    print("رصيدك الحالي =" , balance)
                # 2. السحب النقدي
                elif operation == "2":
                    cash=input("1-100 \n 2-500 \n 3-1000 \n 4-ادخل مبلغ السحب يدويا \n" )
                    if cash=="1" :
                        if balance >= 100:
                             balance -= 100
                             print("تمت العملية بنجاح")
                             print("رصيدك الحالي =", balance)
                        else:
                            print("الرصيد غير كافٍ")

                    elif cash =="2" :
                        if balance >= 500:
                            balance -= 500
                            print( "تمت العملية بنجاح  \n رصيدك الحالي:" , balance )
                        else:
                            print("الرصيد غير كافٍ")


                    elif cash =="3":
                        if balance >= 1000:
                            balance -= 1000
                            print( "تمت العملية بنجاح ورصيدك الحالي: \n" , balance )
                        else:
                            print("الرصيد غير كافٍ")

                    elif cash=="4":
                        manual_cash=int(input("\n :ادخل الرقم الذي تريد سحبه"))
                        if manual_cash <= 0:
                            print("المبلغ غير صحيح")

                        elif manual_cash > balance:
                            print("لقد تحديت الحد الاقصي من رصيدك")
                        elif manual_cash > 10000:
                            print("لقد تجاوزت الحد الأقصى للسحب")
                        else :
                            balance-=manual_cash
                            print( "تمت العملية بنجاح ورصيدك الحالي:" , balance)
                    else:
                        print("اختيار غير صحيح")
                # 3. إيداع الأموال
                elif operation =="3":
                    Deposit=int(input("أدخل المبلغ الذي تريد ايداعه: \n"))
                    if Deposit > 0:
                        balance+=Deposit
                        print("تم ايداع المبلغ بنجاح و رصيدك الحالي : " , balance)
                # 4. الخروج من النظام
                elif operation == "4":
                    print("شكراً لاستخدامك خدماتنا")
                    break
                else:
                    print("اختيار غير صحيح")
            break
        # ==========================================
        # واجهة اللغة الإنجليزية (English Interface)
        # ==========================================
        elif choice == "1":
            print("welcome")

            while True:
                operation = input("Choose the operation you want to perform:\n 1- Check Balance \n 2- Cash Withdrawal \n 3- Deposit \n 4-Exit \n")
                # 1. Check Balance
                if operation == "1":
                    print("Your current balance = " ,balance)
                # 2. Cash Withdrawal
                elif operation == "2":
                    cash = input("1-100 \n 2-500 \n 3-1000 \n 4-Enter manual amount \n")
                    if cash == "1":
                        if balance >= 100:
                            balance -= 100
                            print("the operation was successful and your current balance is:", balance )

                        else:
                            print("Insufficient balance")
                    elif cash == "2":
                        if balance >= 500:
                            balance -= 500
                            print("the operation was successful and your current balance is:", balance)
                        else:
                            print("Insufficient balance")

                    elif cash == "3" :
                        if balance >= 1000:
                            balance -= 1000
                            print("the operation was successful and your current balance is:",balance)
                        else:
                            print("Insufficient balance")
                    elif cash == "4":
                        manual_cash = int(input("\nEnter the amount you want to withdraw: "))
                        if manual_cash <= 0:
                            print("Invalid amount")
                        elif manual_cash > balance:
                            print("You have exceeded your account limit.")
                        elif manual_cash > 10000:
                            print("You have exceeded the maximum withdrawal limit")
                        else:
                            balance -= manual_cash
                            print("The operation was successful \n Your current balance =", balance)
                    else:
                        print("Invalid choice")
                # 3. Deposit
                elif operation == "3":
                    Deposit = int(input("Enter the amount you want to deposit: \n"))
                    if Deposit > 0:
                        balance += Deposit
                        print("Amount deposited successfully \n Your current balance =", balance)
                    else:
                        print("Invalid amount")
                # 4. Exit
                elif operation == "4":
                    print("Thanks for you")
                    break
                else:
                    print("Invalid choice")
            break

        else:
            print("Invalid language choice")

    else:
        print("Wrong card number or password")
# حظر البطاقة عند استنفاد الـ 3 محاولات
else:
    print("Your card is locked")
```
